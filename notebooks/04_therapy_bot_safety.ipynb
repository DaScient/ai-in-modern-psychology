## Setup


```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from code.clinical.therapy_safety import assess_message, build_safety_response
```

## Run the classifier on sample messages


```python
messages = [
    'I had a tough day at work but I made it through.',
    'I keep thinking I want to die — nothing matters anymore.',
    'Sometimes I think about cutting myself when things get hard.',
    'I am taking 800 mg of ibuprofen for a headache.',
    'I just want to end it all, I cannot keep going.',
]
for m in messages:
    r = assess_message(m)
    print(f'INPUT: {m}')
    print(f'  triggered={r.triggered}  severity={r.severity}  patterns={r.matched_patterns}')
    if r.triggered:
        print(f'  RESPONSE: {build_safety_response(r)}')
    print()
```

## Reflect on failure modes

Where would this classifier fail in real conversations?

- Code-mixing or non-English crisis language.
- Metaphorical or musical references ('I want to die laughing').
- Indirect ideation ('I won't be around for next semester').

Production systems combine pattern matching with ML classifiers, conservative thresholds, and human-in-the-loop review.


## Ethical Considerations

- Always test for false-positive harm — flagging a non-crisis message can damage trust.
- Connect detection to **action**: resources and human follow-up are essential.
- Audit performance across demographic groups; crisis-language frequency and form vary by culture.
