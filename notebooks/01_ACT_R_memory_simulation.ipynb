# ACT-R Memory Simulation

**Based on:** *AI in Modern Psychology*, Chapter 6.

This notebook simulates the power-law forgetting curve using a simplified ACT-R declarative-memory module.


## Learning Objectives

- Understand the ACT-R base-level learning equation.
- Visualize the forgetting curve as a function of retrieval history.
- Reflect on the cognitive plausibility of activation-based memory.


## Setup


```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import numpy as np
import matplotlib.pyplot as plt
from code.foundations.act_r_simulation import ACTRMemory
np.random.seed(0)
```

## Build a memory and study it


```python
memory = ACTRMemory(decay=0.5, noise=0.25, threshold=-1.5)
memory.add_chunk('operant_conditioning',   {'domain': 'behaviorism', 'figure': 'Skinner'})
memory.add_chunk('classical_conditioning', {'domain': 'behaviorism', 'figure': 'Pavlov'})

for _ in range(5):
    result = memory.retrieve({'domain': 'behaviorism'}, time_advance=10)
    if result:
        print(f'Retrieved {result.name} at t={memory.current_time:.1f}')
```

## Plot the forgetting curve


```python
times = np.linspace(1, 500, 200)
activations = memory.simulate_forgetting_curve('operant_conditioning', times)

plt.figure(figsize=(8, 4))
plt.plot(times, activations, lw=2, label='Base-level activation')
plt.axhline(memory.threshold, color='r', ls='--', label=f'Threshold ({memory.threshold})')
plt.xlabel('Time since encoding (arbitrary units)')
plt.ylabel('Activation')
plt.title('ACT-R Forgetting Curve (Power Law of Forgetting)')
plt.legend(); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

## Ethical Considerations

Computational memory models are descriptive, not normative. They are useful for designing study schedules and educational interventions, but should not be used to make consequential judgments about individuals (e.g., 'this learner cannot remember').
