## Three sample narratives

```python
narratives = {
    'depressive': (
        "I just feel so hopeless. I can't think clearly anymore. I keep "
        "telling myself I'm worthless. I don't know what I'm doing."
    ),
    'reflective': (
        "We had a difficult week, but we tried to consider what we could "
        "learn. The team felt grateful for the support and chose to think "
        "carefully about next steps."
    ),
    'neutral': (
        "The meeting started at nine and we reviewed the quarterly report. "
        "Three action items were identified for follow-up."
    ),
}
for label, text in narratives.items():
    print(f'--- {label} ---')
    print(extract_features(text).to_dict())
    print()
```

## Language Style Matching

LSM is cosine similarity over function-word vectors and correlates with therapeutic alliance (Niederhoffer & Pennebaker, 2002).

```python
for a in narratives:
    for b in narratives:
        if a < b:
            print(f'LSM({a}, {b}) = {language_style_match(narratives[a], narratives[b]):.3f}')
```
